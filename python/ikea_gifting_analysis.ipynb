```python
import pandas as pd
import numpy as np

df1 = pd.read_csv(
    "online_retail_II page 1.csv",
    encoding="cp1252"
)

df2 = pd.read_csv(
    "online_retail_II page 2.csv",
    encoding="cp1252"
)

df = pd.concat([df1, df2], ignore_index=True)

print("Page 1 shape:", df1.shape)
print("Page 2 shape:", df2.shape)
print("Combined dataset shape:", df.shape)
```

```python
print("df exists:", "df" in globals())
print("df_clean exists:", "df_clean" in globals())
print("df_sales exists:", "df_sales" in globals())
```

```python
# Create a working copy of the combined dataset
df_clean = df.copy()

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# Remove rows with missing product descriptions
df_clean = df_clean.dropna(subset=["Description"])

# Convert invoice dates to datetime
df_clean["InvoiceDate"] = pd.to_datetime(
    df_clean["InvoiceDate"],
    format="%d-%m-%Y %H:%M"
)

# Recalculate revenue
df_clean["Revenue"] = df_clean["Quantity"] * df_clean["Price"]

# Recreate transaction flags
df_clean["Invoice_str"] = df_clean["Invoice"].astype(str)

df_clean["Is_Cancellation"] = (
    df_clean["Invoice_str"].str.startswith("C")
)

df_clean["Is_Bad_Debt_Adjustment"] = (
    df_clean["Description"].str.strip().str.lower()
    == "adjust bad debt"
)

df_clean["Is_Negative_Quantity"] = df_clean["Quantity"] < 0
df_clean["Is_Zero_Price"] = df_clean["Price"] == 0

print("Cleaned dataset shape:", df_clean.shape)
print("Missing descriptions:", df_clean["Description"].isnull().sum())
print("Bad-debt adjustments:", df_clean["Is_Bad_Debt_Adjustment"].sum())
```

```python
# Exclude bad-debt adjustments
df_sales = df_clean[
    ~df_clean["Is_Bad_Debt_Adjustment"]
].copy()

# Keep standard positive-quantity, positive-price sales
df_standard_sales = df_sales[
    (df_sales["Quantity"] > 0) &
    (df_sales["Price"] > 0)
].copy()

print("Cleaned master dataset:", df_clean.shape)
print("Sales dataset:", df_sales.shape)
print("Standard-sales dataset:", df_standard_sales.shape)
```

```python
for data in [df_sales, df_standard_sales]:
    data["Year"] = data["InvoiceDate"].dt.year
    data["Month"] = data["InvoiceDate"].dt.month
    data["Month_Name"] = data["InvoiceDate"].dt.month_name()
    data["Year_Month"] = (
        data["InvoiceDate"].dt.to_period("M").astype(str)
    )

print("Date features recreated successfully.")

print(
    df_standard_sales[
        ["InvoiceDate", "Year", "Month", "Month_Name", "Year_Month"]
    ].head()
)
```

```python
monthly_sales = (
    df_standard_sales
    .groupby("Year_Month")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Unique_Invoices=("Invoice", "nunique")
    )
    .reset_index()
    .sort_values("Year_Month")
)

monthly_sales["Total_Revenue"] = (
    monthly_sales["Total_Revenue"].round(2)
)

display(monthly_sales.head(10))

print("Total months:", len(monthly_sales))
print(
    "Total standard sales revenue:",
    round(monthly_sales["Total_Revenue"].sum(), 2)
)
```

```python
print(monthly_sales.columns.tolist())

display(monthly_sales.head(5).to_string(index=False))

print("\nRevenue total:", round(monthly_sales["Total_Revenue"].sum(), 2))

print(
    "Matches standard-sales revenue:",
    np.isclose(
        monthly_sales["Total_Revenue"].sum(),
        df_standard_sales["Revenue"].sum()
    )
)
```

```python
top_products = (
    df_standard_sales
    .groupby("Description")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("Total_Revenue", ascending=False)
    .head(10)
)

top_products["Total_Revenue"] = top_products["Total_Revenue"].round(2)

display(top_products)
```

```python
top_countries = (
    df_standard_sales
    .groupby("Country")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Unique_Invoices=("Invoice", "nunique")
    )
    .reset_index()
    .sort_values("Total_Revenue", ascending=False)
    .head(10)
)

top_countries["Total_Revenue"] = top_countries["Total_Revenue"].round(2)

display(top_countries)
```

```python
# Export the main cleaned dataset
df_sales.to_csv("cleaned_sales_data.csv", index=False)

# Export positive-quantity, positive-price sales for dashboard analysis
df_standard_sales.to_csv("standard_sales_data.csv", index=False)

# Export monthly summary
monthly_sales.to_csv("monthly_sales_summary.csv", index=False)

# Export top products
top_products.to_csv("top_10_products.csv", index=False)

# Export top countries
top_countries.to_csv("top_10_countries.csv", index=False)

print("All 5 CSV files exported successfully!")
```
